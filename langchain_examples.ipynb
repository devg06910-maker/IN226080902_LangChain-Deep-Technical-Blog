```
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-5-nano")

response = model.invoke("Explain LangChain architecture simply")
print(response.text())
```

```
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template("Explain {topic} in simple terms")

print(prompt.format(topic="LangChain agents"))
```

```
from langchain.chat_models import init_chat_model
from langchain_core.prompts import PromptTemplate

model = init_chat_model("gpt-5-nano")

prompt = PromptTemplate.from_template("Explain the role of {topic} in LangChain")

chain = prompt | model

result = chain.invoke({"topic": "memory"})
print(result.text())
```

```
from langchain.agents import create_agent

def calculator(expression: str):
    return str(eval(expression))

agent = create_agent(
    model="openai:gpt-4.1-mini",
    tools=[calculator],
    system_prompt="You are a helpful assistant"
)

agent.invoke({"messages":[{"role":"user","content":"What is 25 * 4?"}]})
```
